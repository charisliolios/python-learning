```python
temperatures = [12.1, 15.8, 22, 25.6, 32, 18, 42, 33, 40, 10]
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]

average = sum(temperatures) / len(temperatures)

print('average temperature:', average)
print('hottest year:', years[temperatures.index(max(temperatures))])
print('warming over period:', temperatures[-1] - temperatures[0])
```
