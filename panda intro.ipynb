```python
import urllib.request

url = "https://datahub.io/core/global-temp/r/annual.csv"
urllib.request.urlretrieve(url, "global_temp.csv")

print("Downloaded!")
```

```python
f = open("global_temp.csv", "r")
lines = f.readlines()
f.close()

for line in lines[:10]:
    print(line)
```

```python
for line in lines[:3]:
    parts = line.strip().split(",")
    print(parts)
```

```python
data_lines = lines[1:]
```

```python
total_rows = len(data_lines)
print("1. Total rows of data:", total_rows)
```

```python
total_rows = len(data_lines)
highest_anomaly = -999.0  
highest_year = None
lowest_anomaly = 999.0    
lowest_year = None
total_sum = 0.0
positive_count = 0

for line in data_lines:
    parts=line.strip().split(",")
    year=int(parts [1])
    anomaly=float(parts [2])

    if anomaly > highest_anomaly:
        highest_anomaly = anomaly
        highest_year = year
    
    if anomaly < lowest_anomaly:
        lowest_anomaly = anomaly
        lowest_year = year
        
    total_sum += anomaly
    if anomaly > 0:
        positive_count += 1

 
average_anomaly = total_sum / total_rows
print("1. Total rows of data:", total_rows)
print("2. Year with highest anomaly:", highest_year, "(Value:", highest_anomaly, ")")
print("3. Year with lowest anomaly:", lowest_year, "(Value:", lowest_anomaly, ")")
print("4. Average anomaly:", average_anomaly)
print("5. Years with a positive anomaly:", positive_count)

gcag_years = []
gcag_anomalies = []

for line in data_lines:
    parts = line.strip().split(",")
    source = parts[0]
    year = int(parts[1])
    anomaly = float(parts[2])
    
    if source == "GCAG":
        gcag_years.append(year)
        gcag_anomalies.append(anomaly)

print("\nGCAG data only:")
print("First year:", gcag_years[0])
print("Last year:", gcag_years[-1])
print("Number of years:", len(gcag_years))
    
```
